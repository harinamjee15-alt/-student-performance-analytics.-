```python
import pandas as pd
import  numpy as np
data = {
    "Student_ID": list(range(1, 21)),
    "Name": [
        "Rahul", "Aman", "Priya", "Neha", "Rohit",
        "Simran", "Karan", "Anjali", "Vikas", "Pooja",
        "Arjun", "Sneha", "Mohit", "Riya", "Sumit",
        "Kavita", "Deepak", "Nisha", "Varun", "Isha"
    ],
    "Department": ["CSE"] * 20,
    "Python": [80, 65, 90, 45, 70, 85, 55, 95, 60, 75,
               88, 72, 40, 92, 68, 78, 83, 50, 87, 96],
    "Maths": [75, 70, 95, 50, 65, 80, 60, 90, 55, 85,
              82, 68, 45, 94, 72, 76, 88, 52, 79, 91],
    "DBMS": [85, 60, 92, 55, 75, 88, 50, 96, 65, 70,
             90, 74, 42, 89, 66, 81, 84, 58, 86, 93],
    "Attendance": [90, 85, 95, 70, 80, 92, 75, 98, 78, 88,
                   94, 82, 65, 96, 79, 87, 91, 72, 89, 97]
}
df = pd.DataFrame(data)
print(df.to_string(index=False))
df.to_csv("student_data.csv", index=False)
```

```python
#numpy for numerical calculation
subjects = ["Python", "Maths", "DBMS"]
marks = df[subjects].to_numpy()
df["Total"] = np.sum(marks, axis=1)
df["Average"] = np.mean(marks, axis=1)
print(df.to_string(index=False))
df.to_csv("student_data.csv", index=False)
```

```python
#create resuable function for calculation and processing.
def calculate_grade(average):
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 60:
        return "D"
    else:
        return "F" 
df["Grade"] = df["Average"].apply(calculate_grade)
print(df.to_string(index=False)) 
```

```python
#function for pass or fail
def check_result(row):
    for subject in subjects:
        if row[subject] < 40:
            return "Fail"
    return "Pass"

df["Grade"] = df["Average"].apply(calculate_grade)
df["Result"] = df.apply(check_result, axis=1)
print(df.to_string(index=False))
```

```python
print("\nSTUDENT RESULTS")
print(df[[
    "Name", "Total", "Average", "Grade", "Result"
]].to_string(index=False))

```

```python
#highest lowest and overall average performance
print("\nCLASS PERFORMANCE")
print("Highest average:", df["Average"].max())
print("Lowest average:", df["Average"].min())
print("Class average:", df["Average"].mean())
print(df.to_string(index=False))
```

```python
#subject wise performance using numpy.
subject_averages = np.mean(marks, axis=0)
print("\nSUBJECT PERFORMANCE")
for subject, average in zip(subjects, subject_averages):
    print(f"{subject}: {average:.2f}")

best_subject = subjects[np.argmax(subject_averages)]
print("Best-performing subject:", best_subject)
df.to_csv("student_analysis.csv", index=False)
```

```python
#Top performance student
print("\nTOP PERFORMING STUDENT")
top_students = df.nlargest(3, "Average")
print(top_students[["Name", "Average"]].to_string(index=False))
df.to_csv("top_performers.csv", index=False)
```
